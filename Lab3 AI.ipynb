```python
import numpy as np
normal_list=[1,2,3,4,5]
numpy_list=np.array([1,2,3,4,5])
print("Normal python list:",normal_list)
print("Numpy python list:",numpy_list)

#to add any number into each value of numpy list
print("Numpy python list + 2:",numpy_list+2)
#normal list append element not add
print("Normal python list + 2:",normal_list+[2])

```

```python
#create simple numpy list
arr1=np.array([10,20,30,40,50])

#create list 0's of n size
arr2=np.zeros(6)

#create list 1's of n size
arr3=np.ones(5)
#create list of values specific range(1 increment)
arr4=np.arange(1,11)
#create list of values specific range(specific increment)
arr5=np.arange(0,20,2)
print("Arr1:",arr1)
print("Arr2:",arr2)
print("Arr3:",arr3)
print("Arr4:",arr4)
print("Arr5:",arr5)
```

```python
marks=np.array([75,83,91,89,65])
print("First:",marks[0])
print("Third:",marks[2])
print("Last:",marks[-1])
print(marks[1:4])
print(marks[:3])
print(marks[2:])
```

```python
ai_marks=np.array([10,20,30])
maths_marks=np.array([10,20,30])
#adding coresponding values of two arrays
print("Total Marks:",ai_marks+maths_marks)
print("Difference:",ai_marks-maths_marks)
```

```python
marks=np.array([75,83,91,89,65])
print("Sum:",np.sum(marks))
print("Mean:",np.mean(marks))
print("Meadian:",np.median(marks))
print("Min:",np.min(marks))
print("Max:",np.max(marks))
print("Standard Deviation:",np.std(marks))
```

```python
#numpy 2d array
marks=np.array([[10,20,30],
                [40,50,60],
                [70,80,90]])
print(marks)
print(marks[0])
print(marks[0,1])
print(marks[:,0])
```

```python
numbers=np.arange(1,13)
print("Original Array:",numbers)
print("Shape:",numbers.shape)
print("Dimensions:",numbers.ndim)

matrix=numbers.reshape(3,4)
print("Original Array:",matrix)
print("Shape:",matrix.shape)
print("Dimensions:",matrix.ndim)

one_d=matrix.reshape(1,12)
print(one_d)
```

```python
np.random.seed(42) #if you want to same values after runing multiple times
random_values=np.random.rand(5)
random_integers=np.random.randint(1,100,size=5)
print(random_values)
print(random_integers)
```

```python
import pandas as pd
marks=pd.Series([10,20,30,40,50])
print(marks)#print data in form of table
```

```python
marks=pd.Series(
    [75,70,85],
    index=["Ali","Ahmad","Sara"]
)
print(marks)
print("Ali's marks:",marks["Ali"])
```

```python
data={
    "Name":["Ali","Sara","Ahmad","Ayesha","Usman"],
    "AI":[85,75,67,89,90],
    "Programing":[85,75,67,89,90],
    "Maths":[90,45,69,80,99]
}
df=pd.DataFrame(data)
print(df)
print("Shape of Array:",df.shape)
print("columns of table:",df.columns)
print("DataType of table:",df.dtypes)
print("Indormation:",df.info())
```

```python
marks={
    'Name':["Ali","Ahmad","Sara"],
    'Marks':[75,70,85],
    'Age':[10,20,40]
}
df=pd.DataFrame(marks,index=["a","b","c"])
print(df)
print(df.loc["c","Name"])
print(df.loc["a","Marks"])
```
