```python
#task1.2
def sundaes():
    flavors=["vanilla", "chocolate", "strawberry", "pistacchio"] 
    sauces  = ["caramel", "butterscotch", "chocolate"]
    total_combinations=0
    for i in range(len(flavors)):
        for j in range(len(sauces)):
            print(flavors[i] + " ice cream sundae with " + sauces[j] + " sauce")
            total_combinations+=1
    return total_combinations
sundaes()

```

```python
#task1.3
def triangle(): 
    value = 1 
    row = 1 
    while row <= 10: 
        column = 1 
        while column <= row: 
            if column != 10: 
                print(value, ' ', sep = '', end = '') 
            else: 
                print(value) 
            value = value + 1 
            column = column + 1 

        print() 

        row = row + 1

triangle()
```

```python
def count_pattern(pattern,lst):
    if not pattern:
        return 0
    count=0
    pattern_len=len(pattern)
    for i in range(len(lst)-pattern_len+1):
        if lst[i:i+pattern_len]==pattern:
            count+=1
    return count
print(count_pattern(('a','b','a'),('f','a','b','a','f','a','s','a','b','a')))
    
```

```python
#1.4(2)
class RomanNumeral:
    def __init__(self,number):
        self.number=number
    def toRoman(self):
        values = [1000, 900, 500, 400, 100, 90, 50, 40, 10, 9, 5, 4, 1]
        symbols = ["M", "CM", "D", "CD", "C", "XC", "L", "XL", "X", "IX", "V", "IV", "I"]
        roman=""
        for i in range(len(values)):
            while self.number>=values[i]:
                roman=roman+symbols[i]
                self.number=self.number-values[i]
        return roman
number=int(input("Enter A Number:"))
obj=RomanNumeral(number)
print("Roman Numeral:", obj.toRoman())
```

```python
#1.4(3)
class FindPair:
    def __init__(self,target):
        self.target=target
    def pair(self):
        set=[10,20,30,10,40,45,5]
        for i in range(len(set)-1):
            if set[i]+set[i+1]==self.target:
                print("Pair Found:",i," ",i+1)
target=40
obj=FindPair(target)
obj.pair()
```

```python
#1.4(4)
class TripleSum:
    def __init__(self,numbers):
        self.numbers=numbers
    def find_triple(self):
        result=[]
        for i in range(len(self.numbers)):
            for j in range(len(self.numbers)):
                for k in range(len(self.numbers)):
                    if self.numbers[i]+self.numbers[j]+self.numbers[k]==0:
                        result.append([self.numbers[i],
                                      self.numbers[j],
                                      self.numbers[k]])
        return result
numbers = [-25, -10, -7, -3, 2, 4, 8, 10]
obj=TripleSum(numbers)
print(obj.find_triple())
```

```python
#1.4(5)
class ReverseString:
    def __init__(self,line):
        self.line=line
    def reverser(self):
        words=self.line.split()
        reversed_string=""
        for i in range(len(words)-1,-1,-1):
            reversed_string=reversed_string+words[i]+" "
        return reversed_string
line="I Love Pakistan"
obj=ReverseString(line)
print(obj.reverser())
```

```python
#1.4(7)
def matrix_add(m1,m2):
    m3=[[0,0],
        [0,0]]
    for i in range(len(m1)):
        for j in range(len(m2)):
            m3[i][j]=m1[i][j]+m2[i][j]
    return m3
m1=[[1,2],
    [3,4]]
m2=[[4,3],
    [2,1]]
m3=matrix_add(m1,m2)
print(m3)
```

```python
#1.4(8)
def matrix_multiply(m1,m2):
    m3=[[0,0],
        [0,0]]
    for i in range(len(m1)):
        for j in range(len(m2)):
            for k in range(len(m1[0])):
                m3[i][j]=m3[i][j]+m1[i][k]*m2[k][j]
    return m3
m1=[[1,2],
    [3,4]]
m2=[[4,3],
    [2,1]]
m3=matrix_multiply(m1,m2)
print(m3)
```

```python
#1.4(9)
def calculator(num1, num2, operator, d_type="float"):

    if operator == '+':
        if d_type == "int" or d_type == "INT":
            print(num1, "+", num2, "=", int(num1 + num2))
        elif d_type == "float" or d_type == "FLOAT":
            print(num1, "+", num2, "=", float(num1 + num2))
        else:
            print(num1, "+", num2, "=", float(num1 + num2))

    elif operator == '-':
        if d_type == "int" or d_type == "INT":
            print(num1, "-", num2, "=", int(num1 - num2))
        elif d_type == "float" or d_type == "FLOAT":
            print(num1, "-", num2, "=", float(num1 - num2))
        else:
            print(num1, "-", num2, "=", float(num1 - num2))

    elif operator == '*':
        if d_type == "int" or d_type == "INT":
            print(num1, "*", num2, "=", int(num1 * num2))
        elif d_type == "float" or d_type == "FLOAT":
            print(num1, "*", num2, "=", float(num1 * num2))
        else:
            print(num1, "*", num2, "=", float(num1 * num2))

    elif operator == '/':
        if num2 == 0:
            print("Denominator must be greater than zero")
            return
        else:
            print(num1, "/", num2, "=", float(num1 / num2))

    else:
        print("Invalid operator choice")


calculator(4, 2, "+", "INT")
```
