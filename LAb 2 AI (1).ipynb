```python
#task2.2
def merge_sort(elements):
    n=len(elements)
    size=1
    while size<n:
        left=0
        while left<n:
            mid=min(left+size,n)
            right=min(left+size*2,n)

            i=left
            j=mid
            temp=[]
            while i<mid and j<right:
                if elements[i]<=elements[j]:
                    temp.append(elements[i])
                    i=i+1
                else:
                    temp.append(elements[j])
                    j=j+1
            while i<mid:
                temp.append(elements[i])
                i+=1
            while j<right:
                temp.append(elements[j])
                j+=1
            elements[left:right]=temp
            left+=2*size
        size*=2
    return elements
    
elements=[5,1,3,7,8,10,1,2]
print("Merge Sort:",merge_sort(elements))
```

```python
#task2.3
def bubble_sort(elements):
    for i in range(len(elements)):
        for j in range(len(elements)-i-1):
            if elements[j]>elements[j+1]:
                temp=elements[j]
                elements[j]=elements[j+1]
                elements[j+1]=temp
elements=['P','Y','T','H','O','N']
bubble_sort(elements)
print("Buble Sort:",elements)

def insertion_sort(elements):
    n=len(elements)
    for i in range(1,n):
        key=elements[i]
        j=i-1
        while j>=0 and elements[j]>key:
            elements[j+1]=elements[j]
            j=j-1
        elements[j+1]=key
elements=['P','Y','T','H','O','N']
insertion_sort(elements)
print("Insertion Sort:",elements)

def selection_sort(elements):
    n=len(elements)
    for i in range(n):
        location_of_lowest=i
        for j in range(i+1,n):
            if elements[location_of_lowest]>elements[j]:
                location_of_lowest=j
        temp=elements[i]
        elements[i]=elements[location_of_lowest]
        elements[location_of_lowest]=temp;
elements=['P','Y','T','H','O','N']
selection_sort(elements)
print("Selection Sort:",elements)

def merge_sort(elements):
    n=len(elements)
    size=1
    while size<n:
        left=0
        while left<n:
            mid=min(left+size,n)
            right=min(left+size*2,n)

            i=left
            j=mid
            temp=[]
            while i<mid and j<right:
                if elements[i]<=elements[j]:
                    temp.append(elements[i])
                    i=i+1
                else:
                    temp.append(elements[j])
                    j=j+1
            while i<mid:
                temp.append(elements[i])
                i+=1
            while j<right:
                temp.append(elements[j])
                j+=1
            elements[left:right]=temp
            left+=2*size
        size*=2
    return elements
    
elements=['P','Y','T','H','O','N']
print("Merge Sort:",merge_sort(elements))
```

```python
#task2.5
def dup(elements):
    duplicates=[]
    count=0
    for i in range(len(elements)):
        for j in range(i+1,len(elements)):
            if elements[i]==elements[j]:
                duplicates.append(elements[i])
                count=count+1
               
    return duplicates
elements=[1,2,3,4,2,5,6,4]
duplicates=dup(elements)
print("Duplicates: ",duplicates)
```

```python
#task2.9

import time
big_list=list(range(100000,300001))

def search(elements,key):
    for i in range(len(elements)):
        if elements[i]==key:
            return i
        
    return None
def timer(key):
    start=time.time()
    search(big_list,key)
    end=time.time()
    return end-start

print("Front (100000):", timer(100000), "seconds")
print("Middle (200000):", timer(200000), "seconds")
print("Back (300000):", timer(300000), "seconds")
print("Not in list (3):", timer(3), "seconds")
```
