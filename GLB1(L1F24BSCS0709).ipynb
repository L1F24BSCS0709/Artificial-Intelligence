```python
#Task1 1
class NumtoRoman:
    def __init__(self,num):
        self.num=num
    def convertToRoman(self):
        values = [1000, 900, 500, 400, 100, 90, 50, 40, 10, 9, 5, 4, 1]
        symbols = ["M", "CM", "D", "CD", "C", "XC", "L", "XL",
                   "X", "IX", "V", "IV","I"]
        result=""
        for i in range(len(values)):
            while self.num>=values[i]:
                result = result + symbols[i]
                self.num = self.num - values[i]
        return result
num=int(input("Enter Any Value To convert Into Roman"))
obj=NumtoRoman(num)
obj.convertToRoman()
```

```python
#task2
def isAnagram(text1,text2):
    if len(text1)!=len(text2):
        print("These 2 Texts Are Not Anagrams")
        return
    else:
        check=0
        for char1 in text1:
            if char1 not in text2:
                check=0
                break
            else:
                check=1
        if check==1:
            print("Both Words Are Anagrams")
        else:
            print("Both Words Are Not Anagrams")
t1=input("Enter 1st Word:")
t2=input("Enter 2nd Word:")
isAnagram(t1,t2)
```
