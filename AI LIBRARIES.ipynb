```python
import numpy as np
normal_list = [1,2,3,4,5]
numpy_list=np.array([1,2,3,4,5])
print("normal python list:",normal_list)
print("numpy list:",numpy_list)        

print("numpy array + 2:",numpy_list+2)
print("Normal list  + 2:",normal_list+[2])
```

```python
#it will simply create a list and store values
arr1=np.array([10,20,30,40,50])

#it will create list and store zero at each index
arr2=np.zeros(5)

#it will create list and store one at each index
arr3=np.ones(5)

#it will create list and store values from 1 to 10
arr4=np.arange(1,11)

#it will create list and store values from 0 to 20 increment 2
arr5=np.arange(0,20,2)
print("arr 1:",arr1)
print("arr 2:",arr2)
print("arr 3:",arr3)
print("arr 4:",arr4)
print("arr 5:",arr5)
```

```python
marks=np.array([75,83,91,60,65])
#print value of first index
print("first;",marks[0])

#print value of third index
print("third;",marks[2])

#print value at last index
print("last;",marks[-1])

#print values from index 1 to 3
print(marks[1:4])

#print values from 0 index to 2 index
print(marks[:3])

#print values from 2 index to  last 
print(marks[2:])
```

```python
ai_marks=np.array([10,20,30])
maths_marks=np.array([10,20,30])
print("total marks:" ,ai_marks+maths_marks)
print("difference:" ,ai_marks-maths_marks)
```

```python
marks = np.array([1,2,3,4,5])
print("sum:",np.sum(marks))
print("mean:",np.mean(marks))
print("median:",np.median(marks))
print("min:",np.min(marks))
print("max:",np.max(marks))
print("standard deviation:",np.std(marks))
```

```python
marks=np.array([
    [10,20,30],
    [40,50,60],
    [70,80,90]
])
print(marks)
#print first row 
print(marks[0])
#print each rows zero column 1 matrix
print(marks[0,1])
#print each rows zero index
print(marks[:,0])
```

```python
numbers=np.arange(1,13)
print("original array:",numbers)
#tell about shape of list
print("shape:",numbers.shape)
# tell about dimensions
print("dimensions:",numbers.ndim)
#this will use to convert 1d array in 2d by giving rows and columns 
matrix=numbers.reshape(3,4)
print("reshape:",matrix)
numbers2=matrix.reshape(1,12)
print("reshape 2:",numbers2)
```
