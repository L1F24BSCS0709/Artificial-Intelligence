```python
#bfs

def bfs(graph,start,goal):
    cost=0
    path=[]
    visited=set()
    queue=[(start,cost)]
    while queue:
        current,cost=queue.pop(0)
        if current==goal:
            path.append(current)
            return path,cost
        if current not in visited:
            visited.add(current)
            path.append(current)
            for x,y in graph.get(current,{}).items():
                queue.append((x,y+cost))
    return None,float('inf')
graph={
    'A':{'B':1,'C':2},
    'B':{'D':3,'E':6},
    'C':{'F':4,'G':5},
    'D':{},
    'E':{},
    'F':{},
    'G':{}
}
bfs(graph,'A','G')
```

```python
#bfs

def bfs(graph,start,goal):
    cost=0
    path=[]
    visited=set()
    stack=[(start,cost)]
    while stack:
        current,cost=stack.pop()
        if current==goal:
            path.append(current)
            return path,cost
        if current not in visited:
            visited.add(current)
            path.append(current)
            for x,y in graph.get(current,{}).items():
                stack.append((x,y+cost))
    return None,float('inf')
graph={
    'A':{'B':1,'C':2},
    'B':{'D':3,'E':6},
    'C':{'F':4,'G':5},
    'D':{},
    'E':{},
    'F':{},
    'G':{}
}
bfs(graph,'A','G')
```

```python
#ucs --> unform cost search
graph={
    'A':{'B':1,'C':2},
    'B':{'D':3},
    'C':{'D':1},
    'D':{}
}
import heapq
def ucs(graph,start,goal):
    cost=0
    path=[]
    visited=set()
    heap=[(cost,start,path)]
    while heap:
        cost,current,path=heapq.heappop(heap)
        if current==goal:
            path.append(current)
            return path,cost
        if current not in visited:
            visited.add(current)
            path=path+[current]
            for x,y in graph.get(current,{}).items():
                heapq.heappush(heap,(y+cost,x,path))
    return None,float('inif')
ucs(graph,'A','D')        
        
    
```
