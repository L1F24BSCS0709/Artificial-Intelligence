```python
graph={
    'A':{'B':1,'C':3},
    'B':{'D':4,'E':5},
    'C':{'F':3,'G':6},
    'D':{},
    'E':{},
    'F':{},
    'G':{}
}
heuristic={
    'A':8,
    'B':6,
    'C':5,
    'D':4,
    'E':4,
    'F':3,
    'G':0
}
def hill_climbing(start,goal):
    cost=0
    path=[]
    current=start
    while True:
        path.append(current)
        if current==goal:
            return path,cost
        neighbors=graph.get(current,{})
        if not neighbors:
            return None,float('inf')
        best=None
        for x,y in neighbors.items():
            if best==None or heuristic[x]<heuristic[best]:
                best=x
                best_cost=y
        if heuristic[best]>=heuristic[current]:
            return None,float('inf')
        cost=cost+best_cost
        current=best
hill_climbing('A','G')
```
